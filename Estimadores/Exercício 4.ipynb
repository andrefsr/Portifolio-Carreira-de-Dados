```python
import matplotlib.pyplot as plt 
import numpy as np 

theta = 0.9
numero_amostras = 100000
X = np.random.uniform(0,theta,numero_amostras)
```

```python
x = np.linspace(0,theta,500)
pdf = np.ones(len(x))*(1/theta)
plt.figure()
plt.grid(alpha=0.3)
plt.plot(x,pdf,ls='--',color='red',label='PDF teórica')
plt.hist(X,density=True,edgecolor='black',color='skyblue',bins='auto',label=f'Amostras de U(0,{theta:.1f})')
plt.ylabel('PDF')
plt.title(f'{numero_amostras} amostras')
plt.xticks(np.arange(0,theta+theta/10,theta/10))
plt.legend()
plt.show()

media = theta/2
var = (theta**2)/12
print(f'Média teórica = {media:.5f}')
print(f'Variância teórica = {var:.5f}\n')
print(f'Média amostral = {np.mean(X):.5f} - Erro base = {(np.abs(np.mean(X)-media)/media)*100:.2f}%')
print(f'Variância amostral = {np.var(X):.5f} - Erro base = {(np.abs(np.var(X)-var)/var)*100:.2f}%')

### Método dos momentos

A = [[1,1],[-1,1]]
A = np.array(A)

B = [[2*np.mean(X)] , [np.sqrt(12*np.var(X))]]
B = np.array(B)

parametros = np.linalg.solve(A,B)

a = parametros[0]
b = parametros[1]

print('\nSeja X_1, X_2, ..., X_N ~ U(0,b)\n')
#print(f'Método dos momentos - Valor de a = {a[0]:.5f} - Erro = {(np.abs(a[0]-x[0])):.2f}')
print(f'Método dos momentos - Valor de b = {b[0]:.5f} - Erro = {(np.abs(b[0]-x[-1])/x[-1])*100:.2f}%')

### Método da máxima verossimilhança

bvm = np.max(X)
avm = np.min(X)

#print(f'Método da verossimilhança - Valor de a = {avm:.5f} - Erro = {np.abs(avm-x[0]):.2f}')
print(f'Método da verossimilhança - Valor de b = {bvm:.5f} - Erro = {(np.abs(bvm-x[-1])/x[-1])*100:.2f}%')
```
