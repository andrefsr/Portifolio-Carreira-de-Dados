```python
import matplotlib.pyplot as plt 
import numpy as np 

theta = 2.5
numero_amostras = 100000
X = np.random.uniform(theta,theta+1,numero_amostras)
```

```python
x = np.linspace(theta,theta+1,500)
pdf = np.ones(len(x))
plt.figure()
plt.grid(alpha=0.3)
plt.plot(x,pdf,ls='--',color='red',label='PDF teórica')
plt.hist(X,density=True,edgecolor='black',color='skyblue',bins='auto',label=f'Amostras de U({theta},{theta + 1:.1f})')
plt.ylabel('PDF')
plt.title(f'{numero_amostras} amostras')
plt.xticks(np.arange(theta,theta+1 + theta/10,theta/10))
plt.legend()
plt.show()

media = (theta+ theta + 1)/2
var = (1)/12
print(f'Média teórica = {media:.5f}')
print(f'Variância teórica = {var:.5f}\n')
print(f'Média amostral = {np.mean(X):.5f} - Erro base = {(np.abs(np.mean(X)-media)/media)*100:.2f}%')
print(f'Variância amostral = {np.var(X):.5f} - Erro base = {(np.abs(np.var(X)-var)/var)*100:.2f}%')

### Método dos momentos

A = [[1,1],[-1,1]]
A = np.array(A)

B = [[2*np.mean(X)] , [np.sqrt(12*np.var(X))]]
B = np.array(B)

parametros = np.linalg.solve(A,B)

a = parametros[0]

print('\nSeja X_1, X_2, ..., X_N ~ U(theta,theta + 1)\n')
print(f'Método dos momentos - Valor de theta = {a[0]:.5f} - Erro = {(np.abs(a[0]-x[0])/x[0])*100:.2f}%')

### Método da máxima verossimilhança

avm = np.min(X)

print(f'Método da verossimilhança - Valor de theta = {avm:.5f} - Erro = {(np.abs(avm-x[0])/x[0])*100:.2f}%')
```
